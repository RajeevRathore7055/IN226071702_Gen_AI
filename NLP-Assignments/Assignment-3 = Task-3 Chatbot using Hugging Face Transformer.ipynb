## Step 2: Load Pre-trained Model

We use DialoGPT, a conversational model trained by Microsoft.

```python
model_name = "facebook/blenderbot-400M-distill"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

## Step 3: Chatbot Implementation

We build a loop that:
- Takes user input
- Generates response
- Maintains conversation history
- Stops when user types 'exit' or 'quit'

```python
# ==========================================
# Conversational Chatbot using BlenderBot
# ==========================================

from transformers import BlenderbotTokenizer, BlenderbotForConditionalGeneration

# -------------------------------
# Load Model
# -------------------------------
model_name = "facebook/blenderbot-400M-distill"

tokenizer = BlenderbotTokenizer.from_pretrained(model_name)
model = BlenderbotForConditionalGeneration.from_pretrained(model_name)

# -------------------------------
# Start Chat
# -------------------------------
print("Chatbot: Hello! I am your AI assistant. Type 'exit' or 'quit' to stop.\n")

# Conversation loop
while True:

    user_input = input("You: ")

    # Exit condition
    if user_input.lower() in ["exit", "quit"]:
        print("Chatbot: Goodbye! 👋")
        break

    # Encode input
    inputs = tokenizer(user_input, return_tensors="pt")

    # Generate response
    reply_ids = model.generate(
        **inputs,
        max_length=100,
        do_sample=True,
        temperature=0.7
    )

    # Decode response
    response = tokenizer.decode(reply_ids[0], skip_special_tokens=True)

    print("Chatbot:", response)
```

## Working Explanation

1. The tokenizer converts text into numerical tokens.
2. The model processes these tokens using transformer architecture.
3. The chatbot generates a response based on previous conversation context.
4. The loop continues until the user exits.
