# **Task 1: Conceptual Understanding**

### Question 1: What is the difference between "Love" and "love" in NLP?
**Answer:**  
- In NLP, case matters. `"Love"` and `"love"` are considered different tokens. 
Lowercasing is used to standardize text so that semantic meaning is not split unnecessarily.

---

### Question 2: What happens if stopwords are not removed?
**Answer:**  
Stopwords are common words like "the", "is", "and" that do not carry meaningful information. Not removing them may:
- Introduce noise
- Increase vocabulary size unnecessarily
- Reduce the efficiency of NLP models

---

### Question 3: Two real-world scenarios where removing stopwords can be harmful
1. **Sentiment Analysis:** Words like "not" or "no" are stopwords, but removing them changes the sentiment ("I am not happy" → "I am happy").  
2. **Question-Answer Systems:** Stopwords can be critical for understanding questions like "What is the capital of France?" Removing "what" changes the meaning.

---

### Question 4: Explain the difference between stemming and lemmatization
**Answer:**  
- **Stemming:** Reduces words to their root form by chopping off endings (e.g., "running" → "run"). It may be imprecise.  
- **Lemmatization:** Converts words to their dictionary/base form using vocabulary and POS tagging (e.g., "better" → "good"). More accurate than stemming.

# **Task 2: Build Advanced Preprocessing Function**

1) Remove numbers
- Example: "I have 2 dogs" → "I have dogs"
2) Remove extra spaces
- Example: "This is good" → "This is good"
3) Handle repeated characters
- Example: "soooo goooood!!!" → "so good"
4) Convert text to lowercase
- Example: "WOW!!! This is GREAT!!!" → "wow this is great"
5) Remove very short tokens
- Remove words with length ≤ 2
- Exception: keep meaningful words like "no", "not"
6) Remove URLs and email-like patterns
- Example: "Visit http://example.com now" → "visit now"


```python
# import re
# import nltk
# from nltk.tokenize import word_tokenize
# from nltk.corpus import stopwords

# nltk.download('punkt')
# nltk.download('stopwords')
# nltk.download('wordnet')

# # Initialize stopwords
# stop_words = set(stopwords.words('english'))

# # Define preprocess_text function
# def preprocess_text(text):
#     if not text:
#         return [], ""
    
#     # 1. Lowercase
#     text = text.lower()
    
#     # 2. Remove URLs and email-like patterns
#     text = re.sub(r'http\S+|www\S+|https\S+', '', text)
#     text = re.sub(r'\S+@\S+', '', text)
    
#     # 3. Remove numbers
#     text = re.sub(r'\d+', '', text)
    
#     # 4. Handle repeated characters (e.g., "looooved" -> "looved")
#     text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    
#     # 5. Remove extra spaces
#     text = re.sub(r'\s+', ' ', text).strip()
    
#     # Tokenize
#     tokens = word_tokenize(text)
    
#     # 6. Remove very short tokens except 'no', 'not'
#     tokens = [t for t in tokens if len(t) > 2 or t in ('no', 'not')]
    
#     return tokens, ' '.join(tokens)

# ===============================
# Task 2: Advanced NLP Preprocessing Function
# ===============================

import re
import nltk
from nltk.tokenize import word_tokenize

# Download NLTK punkt tokenizer (run once)
nltk.download('punkt')

# ------------------------------
# Function: preprocess_text
# ------------------------------
def preprocess_text(text):
    if not text:  # Handle empty string
        return [], ""
    
    # 1. Convert text to lowercase
    text = text.lower()
    
    # 2. Remove URLs and email-like patterns
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)
    
    # 3. Remove numbers
    text = re.sub(r'\d+', '', text)
    
    # 4. Handle repeated characters (soooo -> so)
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    
    # 5. Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    # 6. Tokenize text
    tokens = word_tokenize(text)
    
    # 7. Remove very short tokens except 'no', 'not'
    tokens = [t for t in tokens if len(t) > 2 or t in ('no', 'not')]
    
    # 8. Return tokens + cleaned sentence
    clean_sentence = ' '.join(tokens)
    return tokens, clean_sentence

# ------------------------------
# Example Test
# ------------------------------
sample_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

# ------------------------------
# Run function and print results
# ------------------------------
for i, sentence in enumerate(sample_sentences, 1):
    tokens, clean_sentence = preprocess_text(sentence)
    print(f"--- Sentence {i} ---")
    print(f"Original: {sentence}")
    print(f"Tokens: {tokens}")
    print(f"Cleaned Sentence: {clean_sentence}\n")
```

# **Task 4: Token Analytics**
1) For each processed sentence, compute:
- Total number of tokens
- Number of unique tokens
- Average token length
2) Analysis Questions
- Which sentence had the most noise?
- Which sentence retained the most meaningful tokens after cleaning?


```python
for sentence in sample_sentences:
    tokens, _ = preprocess_text(sentence)
    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))
    avg_token_len = sum(len(t) for t in tokens) / total_tokens if total_tokens else 0
    
    print(f"Sentence: {sentence}")
    print(f"Total Tokens: {total_tokens}, Unique Tokens: {unique_tokens}, Average Token Length: {avg_token_len:.2f}\n")
```

# **Task 5: Frequency Analysis**
Combine all tokens and use:
from collections import Counter

Identify:
- Top 10 most frequent words
- Top 5 least frequent words


```python
from collections import Counter

all_tokens = []
for sentence in sample_sentences:
    tokens, _ = preprocess_text(sentence)
    all_tokens.extend(tokens)

token_counts = Counter(all_tokens)

print("Top 10 Most Frequent Words:", token_counts.most_common(10))
print("Top 5 Least Frequent Words:", token_counts.most_common()[-5:])
```

# **Task 6: Build Full Pipeline Function**

- def full_pipeline(text_list):
- pass

- Expected Output
{
    "tokens": [...],
    "clean_sentences": [...]
}



```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []
    for text in text_list:
        tokens, clean_text_str = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(clean_text_str)
    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }

pipeline_output = full_pipeline(sample_sentences)
pipeline_output
```
