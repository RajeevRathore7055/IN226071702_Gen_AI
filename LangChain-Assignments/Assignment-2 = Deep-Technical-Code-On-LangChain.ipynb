# **🟢 CELL 3 — LLM (HuggingFaceHub)**

```python
from langchain_community.llms import HuggingFaceHub

llm = HuggingFaceHub(
    repo_id="google/flan-t5-large",
    model_kwargs={"temperature":0.7, "max_length":100}
)

print(llm.invoke("Explain LangChain in simple terms"))
```

# **🟢 CELL 4 — Prompt Template**

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
    "Explain {topic} in simple terms"
)

print(prompt.format(topic="LangChain Agents"))
```

# **🟢 CELL 5 — Chain**

```python
from langchain_core.output_parsers import StrOutputParser

chain = prompt | llm | StrOutputParser()

result = chain.invoke({"topic": "LangChain Memory"})
print(result)
```

# **🟢 CELL 6 — Memory**

```python
from langchain.memory import ConversationBufferMemory

memory = ConversationBufferMemory()

memory.save_context(
    {"input": "Hi, I am learning AI"},
    {"output": "Great! AI is powerful"}
)

memory.save_context(
    {"input": "What am I learning?"},
    {"output": "You are learning AI"}
)

print(memory.load_memory_variables({}))
```

# **🟢 CELL 7 — Tool**

```python
from langchain.tools import Tool

def calculator(expr):
    return str(eval(expr))

tools = [
    Tool(
        name="Calculator",
        func=calculator,
        description="Useful for math calculations"
    )
]
```

# **🟢 CELL 8 — Agent**

```python
from langchain.agents import initialize_agent, AgentType

agent = initialize_agent(
    tools,
    llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)

print(agent.run("What is 25 * 4 + 10?"))
```

# **🟢 CELL 9 — Document Loader**

```python
from langchain_community.document_loaders import TextLoader

with open("sample.txt", "w") as f:
    f.write("LangChain is a framework for building LLM applications.")

loader = TextLoader("sample.txt")
docs = loader.load()

print(docs[0].page_content)
```

# **🟢 CELL 10 — RAG (Vector Store)**

```python
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings()

db = Chroma.from_documents(docs, embeddings)

results = db.similarity_search("What is LangChain?")
print(results[0].page_content)
```
