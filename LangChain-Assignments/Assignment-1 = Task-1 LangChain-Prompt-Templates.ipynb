## ✅ Task 1: Replace Hardcoded Prompts (10%)
**Goal:** Convert a plain f-string function into a reusable LangChain `PromptTemplate`.

````python
# ❌ Old approach using f-string (hardcoded - NOT recommended)
# def explain_topic(topic):
#     return f"Explain {topic} in simple terms for beginners"

# ✅ New approach using LangChain PromptTemplate (reusable & dynamic)
task1_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)

# Function to generate the prompt using the template
def explain_topic(topic):
    """
    Generates a beginner-friendly explanation prompt for any given topic.
    Uses LangChain PromptTemplate instead of f-strings.
    """
    return task1_template.format(topic=topic)

# Testing the template
print("📌 Task 1 Output:")
print(explain_topic("Machine Learning"))
print(explain_topic("Neural Networks"))
print(explain_topic("Python"))
````

## ✅ Task 2: Multi-Input Prompt System (15%)
**Goal:** Build a template that accepts multiple inputs: `topic`, `audience`, and `tone`.

````python
# ✅ Multi-input PromptTemplate with 3 variables
task2_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

# Function to generate multi-input prompts
def multi_input_prompt(topic, audience, tone):
    """
    Generates a customized prompt based on topic, audience, and tone.
    """
    return task2_template.format(topic=topic, audience=audience, tone=tone)

# Test Cases as given in the assignment
test_cases = [
    ("AI", "beginners", "friendly"),
    ("Python", "kids", "fun"),
    ("Deep Learning", "engineers", "technical")
]

print("📌 Task 2 Output:")
for topic, audience, tone in test_cases:
    print(f"➡️  {multi_input_prompt(topic, audience, tone)}")
````

## ✅ Task 3: Prompt Variations Engine (15%)
**Goal:** Create 3 different `PromptTemplate`s for Teaching, Interview, and Storytelling styles.

````python
# ✅ Three different PromptTemplates for different styles

# Template 1: Teaching style
teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

# Template 2: Interview style
interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 questions about {topic}"
)

# Template 3: Storytelling style
storytelling_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)

# Dictionary to map style names to their templates
prompt_variations = {
    "Teaching": teaching_template,
    "Interview": interview_template,
    "Storytelling": storytelling_template
}

# Function to run all variations for a given topic
def run_variations(topic):
    """
    Generates prompts in all 3 styles (Teaching, Interview, Storytelling)
    for the given topic.
    """
    print(f"📌 Task 3 Output for topic: '{topic}'")
    for style, template in prompt_variations.items():
        print(f"\n🔹 [{style}] → {template.format(topic=topic)}")

# Run for the required topic
run_variations("Machine Learning")
````

## ✅ Task 4: ChatPromptTemplate System (15%)
**Goal:** Build a chat-based prompt system using System and Human message roles.

````python
# ✅ Role-based behavior definitions for the system
role_behavior = {
    "teacher": "You are a knowledgeable and patient teacher. Explain concepts clearly and simply.",
    "interviewer": "You are a professional interviewer. Ask insightful questions to assess knowledge.",
    "motivator": "You are an enthusiastic motivator. Inspire and encourage the user about the topic."
}

# Function to build and return ChatPromptTemplate based on role and topic
def build_chat_prompt(role, topic):
    """
    Builds a ChatPromptTemplate with a system role and a user message.
    
    Parameters:
        role  (str): One of 'teacher', 'interviewer', 'motivator'
        topic (str): The topic to discuss
    
    Returns:
        list: Formatted chat messages
    """
    # Validate role
    if role not in role_behavior:
        raise ValueError(f"Invalid role '{role}'. Choose from: {list(role_behavior.keys())}")
    
    # Build the ChatPromptTemplate
    chat_template = ChatPromptTemplate.from_messages([
        ("system", role_behavior[role]),               # System message: defines AI behavior
        ("human", "Please help me understand {topic}") # Human message: dynamic user input
    ])
    
    # Format the template with the provided topic
    formatted_messages = chat_template.format_messages(topic=topic)
    return formatted_messages

# Test: role = 'teacher', topic = 'Neural Networks'
print("📌 Task 4 Output:")
messages = build_chat_prompt("teacher", "Neural Networks")
for msg in messages:
    print(f"\n🔹 [{msg.type.upper()}]: {msg.content}")

# Additional tests
print("\n" + "="*60)
messages2 = build_chat_prompt("interviewer", "Deep Learning")
for msg in messages2:
    print(f"\n🔹 [{msg.type.upper()}]: {msg.content}")

print("\n" + "="*60)
messages3 = build_chat_prompt("motivator", "Python")
for msg in messages3:
    print(f"\n🔹 [{msg.type.upper()}]: {msg.content}")
````

## ✅ Task 5: Input Validation Layer (10%)
**Goal:** Create a `validate_inputs()` function that checks audience and tone values.

````python
# ✅ Allowed values for validation
VALID_AUDIENCES = ["beginner", "intermediate", "expert"]
VALID_TONES     = ["formal", "casual", "fun"]

# Default fallback values
DEFAULT_AUDIENCE = "beginner"
DEFAULT_TONE     = "casual"

def validate_inputs(audience, tone, use_defaults=True):
    """
    Validates audience and tone inputs.
    
    Parameters:
        audience     (str):  The target audience (beginner/intermediate/expert)
        tone         (str):  The tone of explanation (formal/casual/fun)
        use_defaults (bool): If True, assign defaults for invalid values.
                             If False, raise ValueError for invalid values.
    
    Returns:
        tuple: Validated (audience, tone)
    """
    # Convert to lowercase for case-insensitive comparison
    audience = audience.lower().strip()
    tone     = tone.lower().strip()
    
    # Validate audience
    if audience not in VALID_AUDIENCES:
        if use_defaults:
            print(f"⚠️  Invalid audience '{audience}'. Defaulting to '{DEFAULT_AUDIENCE}'.")
            audience = DEFAULT_AUDIENCE
        else:
            raise ValueError(f"❌ Invalid audience '{audience}'. Must be one of: {VALID_AUDIENCES}")
    
    # Validate tone
    if tone not in VALID_TONES:
        if use_defaults:
            print(f"⚠️  Invalid tone '{tone}'. Defaulting to '{DEFAULT_TONE}'.")
            tone = DEFAULT_TONE
        else:
            raise ValueError(f"❌ Invalid tone '{tone}'. Must be one of: {VALID_TONES}")
    
    print(f"✅ Valid inputs → Audience: '{audience}', Tone: '{tone}'")
    return audience, tone


# Test Cases
print("📌 Task 5 Output:")
print("\n--- Test 1: Valid inputs ---")
validate_inputs("beginner", "fun")

print("\n--- Test 2: Invalid audience, valid tone (with defaults) ---")
validate_inputs("student", "formal")

print("\n--- Test 3: Valid audience, invalid tone (with defaults) ---")
validate_inputs("expert", "angry")

print("\n--- Test 4: Raise error on invalid input ---")
try:
    validate_inputs("kid", "happy", use_defaults=False)
except ValueError as e:
    print(e)
````

## ✅ Task 6: Prompt Generator App (15%)
**Goal:** Create a `generate_prompt()` function that combines validation, templates, and styles.

````python
# ✅ Style-based PromptTemplates for the generator app
style_templates = {
    "teaching": PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template="Explain {topic} for {audience} in a {tone} teaching style, step by step."
    ),
    "interview": PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template="Ask 3 {tone} interview questions about {topic} for {audience} level candidates."
    ),
    "storytelling": PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template="Explain {topic} for {audience} in a {tone} storytelling style."
    )
}

def generate_prompt(topic, audience, tone, style):
    """
    Main Prompt Generator App.
    
    Combines:
      1. Input Validation (Task 5)
      2. Style-based PromptTemplate selection (Task 3)
      3. Dynamic prompt generation
    
    Parameters:
        topic    (str): Topic to explain
        audience (str): Target audience (beginner/intermediate/expert)
        tone     (str): Tone of explanation (formal/casual/fun)
        style    (str): Style of prompt (teaching/interview/storytelling)
    
    Returns:
        str: The fully generated prompt
    """
    # Step 1: Validate inputs
    audience, tone = validate_inputs(audience, tone)
    
    # Step 2: Validate style
    style = style.lower().strip()
    if style not in style_templates:
        raise ValueError(f"❌ Invalid style '{style}'. Choose from: {list(style_templates.keys())}")
    
    # Step 3: Select the right template and generate the prompt
    selected_template = style_templates[style]
    prompt = selected_template.format(topic=topic, audience=audience, tone=tone)
    
    return prompt


# Test the Prompt Generator App
print("📌 Task 6 Output:")
print("\n--- Example 1: Neural Networks, beginners, fun, storytelling ---")
result = generate_prompt("Neural Networks", "beginners", "fun", "storytelling")
print(f"\n🎯 Generated Prompt:\n{result}")

print("\n" + "="*60)
print("\n--- Example 2: Python, expert, formal, interview ---")
result2 = generate_prompt("Python", "expert", "formal", "interview")
print(f"\n🎯 Generated Prompt:\n{result2}")

print("\n" + "="*60)
print("\n--- Example 3: Deep Learning, intermediate, casual, teaching ---")
result3 = generate_prompt("Deep Learning", "intermediate", "casual", "teaching")
print(f"\n🎯 Generated Prompt:\n{result3}")
````

## ✅ Task 7: Template Reusability Test (10%)

````python
# ✅ ONE reusable template used for 5 different inputs
reusable_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} and engaging way."
)

# 5 different inputs to demonstrate reusability
test_inputs = [
    {"topic": "Artificial Intelligence",  "audience": "beginners",     "tone": "friendly"},
    {"topic": "Data Structures",          "audience": "students",      "tone": "simple"},
    {"topic": "Cloud Computing",          "audience": "professionals", "tone": "formal"},
    {"topic": "Blockchain",               "audience": "kids",          "tone": "fun"},
    {"topic": "Natural Language Processing", "audience": "engineers", "tone": "technical"}
]

print("📌 Task 7 Output – Reusability Test (1 Template, 5 Different Inputs):")
print("=" * 65)

for i, inputs in enumerate(test_inputs, start=1):
    # Same template, different inputs each time
    prompt = reusable_template.format(**inputs)
    print(f"\n✅ Test {i}: {prompt}")

print("\n" + "="*65)
print("\n💡 Observation: Same template structure → 5 different, unique outputs.")
print("   This proves the REUSABILITY of LangChain PromptTemplates!")
````
